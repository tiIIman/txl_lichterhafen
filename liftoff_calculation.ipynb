```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# manual params

liftoffSpeed_kmh = 350
liftoffDistance_m = 1700
```

```python
# derived params
liftoffSpeed_ms = liftoffSpeed_kmh / 3.6
liftoffTime_s = 2*liftoffDistance_m / liftoffSpeed_ms

acceleration = liftoffSpeed_ms / liftoffTime_s
```

```python
# Generate arrays for plotting
timeArr = np.arange(0, liftoffTime_s*1.2, 0.1)
speedArr = acceleration * timeArr
distanceArr = 1/2 * acceleration * timeArr**2
```

```python
# plot both graphs
plt.subplot(2, 1, 1)
plt.plot(timeArr, speedArr)
plt.ylabel('Speed (m/s)')
plt.title('Liftoff Profile')
plt.hlines(liftoffSpeed_ms, 0, liftoffTime_s, color='red', linestyle='--', label='Liftoff Speed')
plt.vlines(liftoffTime_s, 0, liftoffSpeed_ms, color='red', linestyle='--', label='Liftoff Time')

plt.subplot(2, 1, 2)
plt.plot(timeArr, distanceArr)
plt.xlabel('Time (s)')
plt.ylabel('Distance (m)')
plt.hlines(liftoffDistance_m, 0, liftoffTime_s, color='red', linestyle='--', label='Liftoff Distance')
plt.vlines(liftoffTime_s, 0, liftoffDistance_m, color='red', linestyle='--', label='Liftoff Time')

# --- Subüberschrift über beide Plots ---
plt.suptitle(
    f"Choosen Parameters:\n"
    f"Liftoff Speed = {liftoffSpeed_ms:.1f} m/s\n"
    f"Liftoff Distance = {liftoffDistance_m} m\n\n"
    f"Calculated Acceleration = {acceleration:.2f} m/s²\n",
    fontsize=12,
    y=1.03,
    ha="left",
    x=0.0
)

plt.tight_layout()
plt.show()
```
